```python
import pandas as pd
import numpy as np
import gsw

# ======================================================
# 1. 读取并整理站点数据
# ======================================================
df = pd.read_csv(configured_path('bats_profiles_file'))

df = df.rename(columns={
    "Depth": "depth",
    "Temp_x": "temp",
    "Sal1_x": "sal"
})

df[["depth", "temp", "sal"]] = df[["depth", "temp", "sal"]].astype(float)

# ======================================================
# 2. 单剖面 MLD（严格复刻你原定义）
# ======================================================
def calc_mld_sigma0_station(
    depth, temp, sal,
    lat=31.67, lon=295.83,
    ref_depth=10.0,
    threshold=0.03
):
    mask = np.isfinite(depth) & np.isfinite(temp) & np.isfinite(sal)
    depth = depth[mask]
    temp  = temp[mask]
    sal   = sal[mask]

    if len(depth) < 3:
        return np.nan

    idx = np.argsort(depth)
    depth = depth[idx]
    temp  = temp[idx]
    sal   = sal[idx]

    if ref_depth < depth.min() or ref_depth > depth.max():
        return np.nan

    # 压力
    p = gsw.p_from_z(-depth, lat)

    # σ0
    SA = gsw.SA_from_SP(sal, p, lon, lat)
    sigma0 = gsw.sigma0(SA, temp)

    # 10 m 参考层（插值）
    sigma0_ref = np.interp(ref_depth, depth, sigma0)

    delta = np.abs(sigma0 - sigma0_ref)

    valid = (depth > ref_depth) & (delta >= threshold)
    if not np.any(valid):
        return np.nan

    i2 = np.where(valid)[0][0]
    i1 = i2 - 1
    if i1 < 0:
        return np.nan

    d1, d2 = depth[i1], depth[i2]
    r1, r2 = delta[i1], delta[i2]

    if r2 == r1:
        mld = d2
    else:
        frac = (threshold - r1) / (r2 - r1)
        mld = d1 + frac * (d2 - d1)

    # 新增条件：必须大于 ref_depth
    if mld <= ref_depth:
        return np.nan

    return mld


# ======================================================
# 3. 按日期循环，计算完整时间序列
# ======================================================
records = []

for date, g in df.groupby("yyyymmdd"):
    mld = calc_mld_sigma0_station(
        g["depth"].values,
        g["temp"].values,
        g["sal"].values
    )

    records.append({
        "yyyymmdd": pd.to_datetime(str(date)),
        "MLD_sigma0_003": mld
    })

mld_df = pd.DataFrame(records).sort_values("yyyymmdd")

print(mld_df.head())
```

```python
def four_neighbor_stats(ds: xr.Dataset, var_name: str, target_lat, target_lon):
    lat = ds.lat.values
    lon = ds.lon.values

    # 找到距离目标点最近网格索引
    ilat = np.argmin(np.abs(lat - target_lat))
    ilon = np.argmin(np.abs(lon - target_lon))

    # 取左右/上下各一格，组成2×2四点
    lat_inds = []
    for di in (-1, 0):
        idx = ilat + di
        if 0 <= idx < len(lat):
            lat_inds.append(idx)

    lon_inds = []
    for dj in (-1, 0):
        idx = ilon + dj
        if 0 <= idx < len(lon):
            lon_inds.append(idx)

    # 截取2×2子区域，保留time维度
    sub_da = ds[var_name].isel(lat=lat_inds, lon=lon_inds)

    # 四点均值（原需求）
    mean_da = sub_da.mean(dim=["lat", "lon"], skipna=True)
    # 四点标准差（用于shading误差填色）
    std_da = sub_da.std(dim=["lat", "lon"], skipna=True)

    # 统一赋值目标经纬度坐标，和interp输出格式对齐
    mean_da = mean_da.assign_coords(lat=target_lat, lon=target_lon)
    std_da = std_da.assign_coords(lat=target_lat, lon=target_lon)

    return mean_da, std_da
```

```python
def compute_mld_sigma0_vectorized(ds, temp_name='thetao_cglo', salt_name='so_cglo',
                                  depth_name='depth', lat_name='lat', lon_name='lon',
                                  threshold=0.03, ref_depth=10, show_progress=True):
    """
    向量化计算混合层深度 MLD (|Δσ₀|=threshold kg/m³, 参考深度 ref_depth)
    支持数据维度: (time, depth, lat, lon)
    第一个超过阈值的深度必须 > ref_depth
    """
    temp = ds[temp_name].values  # (time, depth, lat, lon)
    salt = ds[salt_name].values
    depth = ds[depth_name].values
    lat = ds[lat_name].values
    lon = ds[lon_name].values
    time = ds["time"].values

    ntime, ndepth, nlat, nlon = temp.shape
    npoints = nlat * nlon

    # 展平空间维度
    theta_flat = temp.reshape(ntime, ndepth, npoints)
    salt_flat  = salt.reshape(ntime, ndepth, npoints)
    lat_flat   = np.repeat(lat, nlon)
    lon_flat   = np.tile(lon, nlat)

    mld = np.full((ntime, npoints), np.nan)

    iterator = range(ntime)
    if show_progress:
        iterator = tqdm(iterator, desc="Computing MLD (daily)")

    for t in iterator:
        theta_t = theta_flat[t]  # (depth, points)
        salt_t  = salt_flat[t]

        # 掩码有效数据
        mask = np.isfinite(theta_t) & np.isfinite(salt_t)
        valid_points = np.any(mask, axis=0)
        if not np.any(valid_points):
            continue

        theta_valid = np.where(mask, theta_t, np.nan)
        salt_valid  = np.where(mask, salt_t, np.nan)

        # 批量计算压力、盐度、密度
        p = gsw.p_from_z(-depth[:, None], lat_flat[None, :])
        SA = gsw.SA_from_SP(salt_valid, p, lon_flat[None, :], lat_flat[None, :])
        sigma0 = gsw.sigma0(SA, theta_valid)  # (depth, points)

        # 参考深度密度 (每个剖面单独插值)
        sigma0_ref = np.apply_along_axis(
            lambda prof: np.interp(ref_depth, depth, prof, left=np.nan, right=np.nan),
            axis=0,
            arr=sigma0
        )

        delta = np.abs(sigma0 - sigma0_ref[None, :])

        # 找到超过阈值且深度 > ref_depth 的第一个索引
        over_mask = (delta >= threshold) & (depth[:, None] > ref_depth)
        over_idx = np.argmax(over_mask, axis=0)
        over_idx[np.all(over_mask == False, axis=0)] = -1  # 没有满足条件的点标记为 -1

        # 批量线性插值
        d1 = np.where(over_idx>0, depth[over_idx-1], np.nan)
        d2 = np.where(over_idx>0, depth[over_idx], np.nan)
        r1 = np.where(over_idx>0, delta[over_idx-1, np.arange(npoints)], np.nan)
        r2 = np.where(over_idx>0, delta[over_idx, np.arange(npoints)], np.nan)
        frac = (threshold - r1) / (r2 - r1)
        mld[t] = d1 + frac*(d2 - d1)
        mld[t, over_idx==-1] = np.nan

    # reshape 回 (time, lat, lon)
    mld = mld.reshape(ntime, nlat, nlon)
    mld_da = xr.DataArray(
        mld,
        coords={"time": time, "lat": lat, "lon": lon},
        dims=("time", "lat", "lon"),
        name=f"MLD_sigma0_{threshold*1000:03.0f}",
        attrs={
            "long_name": f"Mixed Layer Depth (|Δσ₀|={threshold} kg/m³, ref={ref_depth} m)",
            "units": "m",
            "method": "TEOS-10 σ₀ threshold",
        }
    )
    return mld_da
```

```python
import numpy as np
import xarray as xr
from tqdm import tqdm

def compute_mld_from_density(ds, density_name='sigma0',
                             depth_name='depth', lat_name='lat', lon_name='lon',
                             threshold=0.03, ref_depth=10, show_progress=True):
    """
    计算混合层深度 MLD，基于已有密度数据 sigma0，阈值为 threshold kg/m³，参考深度为 ref_depth
    输入数据维度：(time, depth, lat, lon)
    """
    density = ds[density_name].values  # (time, depth, lat, lon)
    depth = ds[depth_name].values
    lat = ds[lat_name].values
    lon = ds[lon_name].values
    time = ds["time"].values

    ntime, ndepth, nlat, nlon = density.shape
    npoints = nlat * nlon

    density_flat = density.reshape(ntime, ndepth, npoints)
    mld = np.full((ntime, npoints), np.nan)

    iterator = range(ntime)
    if show_progress:
        iterator = tqdm(iterator, desc="Computing MLD from density")

    for t in iterator:
        dens_t = density_flat[t]  # (depth, points)

        # 有效数据掩码（排除nan）
        mask = np.isfinite(dens_t)
        valid_points = np.any(mask, axis=0)
        if not np.any(valid_points):
            continue

        dens_valid = np.where(mask, dens_t, np.nan)

        # 参考深度密度，插值获取
        rho_ref = np.apply_along_axis(
            lambda prof: np.interp(ref_depth, depth, prof, left=np.nan, right=np.nan),
            axis=0,
            arr=dens_valid
        )

        delta = np.abs(dens_valid - rho_ref[None, :])

        over_mask = (delta >= threshold) & (depth[:, None] > ref_depth)
        over_idx = np.argmax(over_mask, axis=0)
        no_mld = ~np.any(over_mask, axis=0)
        over_idx[no_mld] = -1

        d1 = np.where(over_idx > 0, depth[over_idx - 1], np.nan)
        d2 = np.where(over_idx > 0, depth[over_idx], np.nan)
        r1 = np.where(over_idx > 0, delta[over_idx - 1, np.arange(npoints)], np.nan)
        r2 = np.where(over_idx > 0, delta[over_idx, np.arange(npoints)], np.nan)

        frac = (threshold - r1) / (r2 - r1)
        mld[t] = d1 + frac * (d2 - d1)

        mld[t, over_idx == -1] = np.nan

    mld = mld.reshape(ntime, nlat, nlon)

    mld_da = xr.DataArray(
        mld,
        coords={"time": time, "lat": lat, "lon": lon},
        dims=("time", "lat", "lon"),
        name=f"MLD_sigma0_{int(threshold*1000):03d}",
        attrs={
            "long_name": f"Mixed Layer Depth (|Δσ₀|={threshold} kg/m³, ref={ref_depth} m)",
            "units": "m",
            "method": "Density threshold method",
        }
    )
    return mld_da
```

```python
# Prepare SOTS and PIRATA monthly station data for the final comparison figure
SOTS_FILE = configured_path('sots_mld_file')
PIRATA_FILE = configured_path('pirata_mld_file')
PRED_FILE = configured_path('paired_mld_file')

pred_ds = xr.open_dataset(PRED_FILE)
pred_time = pd.date_range('1982-01-01', periods=pred_ds.sizes['time'], freq='MS') + pd.Timedelta(days=14)
if pred_ds.sizes['time'] != len(pred_time):
    raise ValueError(f'Unexpected prediction time length: {pred_ds.sizes["time"]}')
pred_ds = pred_ds.assign_coords(time=pred_time)

def prepare_monthly_station(station_file, target_lat, target_lon):
    station_ds = xr.open_dataset(station_file)
    obs = station_ds['MLD_monthly_mean'].sel(time=slice('2010-01', '2020-12')).load()
    # Normalize every monthly timestamp to the 15th so station and model months align exactly.
    obs = obs.assign_coords(time=pd.DatetimeIndex(obs.time.values).to_period('M').to_timestamp() + pd.Timedelta(days=14))
    pred, pred_std = four_neighbor_stats(pred_ds, 'MLD_pred', target_lat, target_lon)
    pred = pred.sel(time=slice('2010-01', '2020-12')).load()
    pred_std = pred_std.sel(time=slice('2010-01', '2020-12')).load()
    obs, pred, pred_std = xr.align(obs, pred, pred_std, join='inner')
    # Keep the complete monthly axis. Mask model values wherever observations are missing,
    # so gaps remain discontinuous and model/observation comparisons use the same months.
    obs_valid = np.isfinite(obs)
    pred = pred.where(obs_valid)
    pred_std = pred_std.where(obs_valid)
    station_ds.close()
    return obs.values, pred.values, pred_std.values, obs.time.values

# SOTS moorings are near 47°S, 142°E; PIRATA coordinates are stored in its NetCDF file.
da1_clean_5, da2_clean_5, std_clean_5, times_clean_5 = prepare_monthly_station(SOTS_FILE, -47.0, 142.0)
with xr.open_dataset(PIRATA_FILE) as pirata_meta:
    pirata_lat = float(pirata_meta['lat'].values)
    pirata_lon = float(pirata_meta['lon'].values) % 360
da1_clean_6, da2_clean_6, std_clean_6, times_clean_6 = prepare_monthly_station(PIRATA_FILE, pirata_lat, pirata_lon)

print(f'SOTS paired months: {np.count_nonzero(np.isfinite(da1_clean_5) & np.isfinite(da2_clean_5))}')
print(f'PIRATA paired months: {np.count_nonzero(np.isfinite(da1_clean_6) & np.isfinite(da2_clean_6))}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score

import matplotlib as mpl
from matplotlib import font_manager
from matplotlib.offsetbox import AnchoredText
from pathlib import Path

times_font_dir = Path(configured_path('font_dir'))
times_font_files = [times_font_dir / 'Times New Roman.ttf', times_font_dir / 'Times New Roman Bold.ttf']
for font_file in times_font_files:
    if not font_file.is_file():
        raise FileNotFoundError(f'Required Times New Roman font not found: {font_file}')
    font_manager.fontManager.addfont(str(font_file))
resolved_times = Path(font_manager.findfont('Times New Roman', fallback_to_default=False)).resolve()
if resolved_times not in {font.resolve() for font in times_font_files}:
    raise RuntimeError(f'Times New Roman resolved to an unexpected file: {resolved_times}')
mpl.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'mathtext.fontset': 'custom',
    'mathtext.rm': 'Times New Roman',
    'mathtext.it': 'Times New Roman:italic',
    'mathtext.bf': 'Times New Roman:bold',
    'axes.unicode_minus': False,
})

def mbe(true, pred):
    return np.mean(pred - true)

# Six stations: observations, SJTU-ViT predictions, paired times, and four-grid-point spread.
da1_clean_list = [da1_clean_1, da1_clean_2, da1_clean_3, da1_clean_4, da1_clean_5, da1_clean_6]
da2_clean_list = [da2_clean_1, da2_clean_2, da2_clean_3, da2_clean_4, da2_clean_5, da2_clean_6]
times_clean_list = [times_clean_1, times_clean_2, times_clean_3, times_clean_4, times_clean_5, times_clean_6]
std_clean_list = [std_clean_1, std_clean_2, std_clean_3, std_clean_4, std_clean_5, std_clean_6]

titles = ['BAT (31.67°N, 295.83°E)', 'TAO (8°N, 137°E)', 'HOT (22.75°N, 202°E)',
          'Papa (50.13°N, 215.17°E)', 'SOTS (47°S, 142°E)', 'PIRATA (10°S, 350°E)']
labels = ['BAT', 'TAO', 'HOT', 'Papa', 'SOTS', 'PIRATA']
subplot_labels = ['(a)', '(b)', '(c)', '(d)', '(e)', '(f)']

fig, axes = plt.subplots(2, 3, figsize=(24, 11), sharex=True)
axes = axes.flatten()

for i in range(6):
    true = da1_clean_list[i]
    pred = da2_clean_list[i]
    times = times_clean_list[i]
    std  = std_clean_list[i]  # 取出当前站点标准差

    # Mask predictions with observations for plotting, then use only paired finite values
    # for MBE, RMSE, and R². NaNs remain in place so missing months break the lines.
    obs_valid = np.isfinite(true)
    pred_plot = np.where(obs_valid, pred, np.nan)
    std_plot = np.where(obs_valid, std, np.nan)
    paired = obs_valid & np.isfinite(pred_plot)
    true_paired = true[paired]
    pred_paired = pred_plot[paired]
    if true_paired.size < 2:
        raise ValueError(f'Not enough paired observations and predictions for {labels[i]}')

    mbe_val = mbe(true_paired, pred_paired)
    rmse_val = np.sqrt(mean_squared_error(true_paired, pred_paired))
    r2_val = r2_score(true_paired, pred_paired)

    ax = axes[i]
    ax.plot(times, pred_plot, label='SJTU-ViT', linestyle='-', linewidth=2.7, color="#d62728")
    ax.plot(times, true, label=labels[i], linestyle='-', linewidth=2.7, color="#1f77b4")    
    
    # 新增：四点标准差阴影shading
    ax.fill_between(times,
                    pred_plot - std_plot,
                    pred_plot + std_plot,
                    alpha=0.15, color='#d62728')

    # Only the second row shows x-axis labels.
    if i >= 3:
        ax.set_xlabel('Date (YYYY)', fontsize=20)
    # Only the first column shows y-axis labels.
    if i in [0, 3]:
        ax.set_ylabel('MLD (m)', fontsize=20)

    if i < 3:
        ax.tick_params(axis='x', labelbottom=False)
        
    ax.set_title(titles[i], fontsize=21)

    ax.tick_params(axis='x', labelsize=20)
    ax.tick_params(axis='y', labelsize=20)

    # ========== 1. 指标改为单行，横向分隔 ==========
    textstr = f'MBE = {mbe_val:.2f} RMSE = {rmse_val:.2f} R² = {r2_val:.2f}'
    metrics_box = AnchoredText(textstr, loc='upper left', prop={'size': 17.5},
                               frameon=True, borderpad=0.5, pad=0.3)
    metrics_box.patch.set_boxstyle('round,pad=0.2')
    metrics_box.patch.set_facecolor('white')
    metrics_box.patch.set_alpha(0.8)
    ax.add_artist(metrics_box)

    ax.text(0.02, 1.07, subplot_labels[i], transform=ax.transAxes,
            fontsize=21,  va='top', ha='left')

    # ========== 2. 图例单行 ncol=3，右上角不挡文字 ==========
    ax.legend(loc='upper right', fontsize=17.5, ncol=2, columnspacing=0.7,
              handletextpad=0.3, handlelength=0.9)

    # ========== 3. 向上扩大y轴范围，预留顶部空白，彻底不遮挡线条 ==========
    y_min, y_max = ax.get_ylim()
    if i < 3:
        # Top row: slightly expand the upper limit.
        scale = 1.03
    else:
        # Bottom row: leave more room above the series.
        scale = 1.14
    ax.set_ylim(y_min, y_max * scale)

plt.tight_layout()
output_file = Path(str(Path(configured_path('output_dir')) / 'figure4.png'))
output_file.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(output_file, dpi=600, bbox_inches='tight', pad_inches=0)
print(f'Saved: {output_file}')
plt.show()
```
